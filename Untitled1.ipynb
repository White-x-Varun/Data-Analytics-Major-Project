```python
import numpy as np
import pandas as pd

# Load data - corrected to read_excel and specified the 'DATA' sheet
file_path = r'C:\Users\JAISHDIWAKAR\Downloads\IIOT Major Project.xlsx'
df_data = pd.read_excel(file_path, sheet_name='DATA')

# Convert columns to NumPy arrays
sales = df_data['Sales'].to_numpy()
profit = df_data['Profit'].to_numpy()
quantity = df_data['Quantity'].to_numpy()
order_status = df_data['Order_Status'].to_numpy()

# 1. Basic array mathematics
total_sales = np.sum(sales)
average_quantity = np.mean(quantity)
max_profit = np.max(profit)

# 2. Conditional selection using np.where (e.g., mark high-value sales > 5000)
high_value_orders = np.where(sales > 5000, "High Value", "Standard")

# 3. Calculate Profit Margin percentage using numpy safety division to avoid 0s
profit_margin = np.divide(profit, sales, out=np.zeros_like(profit), where=sales!=0) * 100

# Print Results
print(f"Total Sales: {np.round(total_sales, 2)}")
print(f"Average Items Per Order: {np.round(average_quantity, 2)}")
print(f"Max Single Profit: {np.round(max_profit, 2)}")
print(f"Average Profit Margin: {np.round(np.mean(profit_margin), 2)}%")
```

```python
import pandas as pd

# 1. Load the dataset directly from your local Excel file's 'DATA' sheet
# Using 'r' before the string to safely handle Windows file path backslashes
file_path = r'C:\Users\JAISHDIWAKAR\Downloads\IIOT Major Project.xlsx'
df = pd.read_excel(file_path, sheet_name='DATA')

# 2. Filter data (e.g., only Successful/Delivered Orders)
delivered_df = df[df['Order_Status'] == 'Delivered']

# 3. Aggregation & Grouping (Total Sales and Profit by Category)
category_summary = delivered_df.groupby('Category')[['Sales', 'Profit']].agg(['sum', 'mean'])

print("--- Sales & Profit by Category ---")
print(category_summary)

# 4. Value Counts (Frequency of Payment Modes)
payment_distribution = df['Payment_Mode'].value_counts()

print("\n--- Distribution of Payment Modes ---")
print(payment_distribution)

# 5. Sorting (Top 5 orders with the highest profit)
top_profitable_orders = df.nlargest(5, 'Profit')[['Order_ID', 'Customer_Name', 'Category', 'Profit']]

print("\n--- Top 5 Most Profitable Orders ---")
print(top_profitable_orders)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Load the dataset directly from your local Excel file's 'DATA' sheet
file_path = r'C:\Users\JAISHDIWAKAR\Downloads\IIOT Major Project.xlsx'
df = pd.read_excel(file_path, sheet_name='DATA')

# Prepare data
# Filter out invalid orders and group by month name
monthly_sales = df[df['Valid_Order'] == 1].groupby('Month_Name')['Sales'].sum()
ordered_months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
monthly_sales = monthly_sales.reindex(ordered_months)

# Create a clean line plot
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='firebrick', linestyle='-', linewidth=2)

# Customize elements using Matplotlib
plt.title('Monthly Sales Trend (Jan - May 2025)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Sales (INR)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. Load the dataset directly from your local Excel file's 'DATA' sheet
file_path = r'C:\Users\JAISHDIWAKAR\Downloads\IIOT Major Project.xlsx'
df = pd.read_excel(file_path, sheet_name='DATA')

# Set aesthetic style for Seaborn
sns.set_theme(style="whitegrid")

# ==========================================
# PLOT 1: Sales Breakdown by Category & Gender
# ==========================================
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df[df['Valid_Order'] == 1],
    x='Category',
    y='Sales',
    hue='Gender',
    estimator=sum,
    errorbar=None,
    palette='muted'
)

plt.title('Sales Breakdown by Category and Gender', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Sales (INR)', fontsize=12)
plt.legend(title='Gender')

plt.tight_layout()
plt.show()

# ==========================================
# PLOT 2: Heatmap (Correlation Matrix)
# ==========================================
plt.figure(figsize=(8, 6))

# Filter for the key numerical columns to check correlations
numeric_cols = df[['Sales', 'Profit', 'Quantity', 'Rating', 'Net_Profit']]

sns.heatmap(
    numeric_cols.corr(), 
    annot=True, 
    cmap='coolwarm', 
    fmt=".2f", 
    linewidths=0.5
)

plt.title('Correlation Matrix of Metrics', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
